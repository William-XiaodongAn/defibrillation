```python
import numpy as np

# 1. Set the grid dimensions (1000 x 1000)
nx, ny = 1024, 1024

# 2. Define uniform field components
# For a standard uniform field pointing in the x-direction:
# Ex = 1.0, Ey = 0.0, Magnitude = 1.0, Angle = 0.0
Ex = np.ones((nx, ny))
Ey = np.zeros((nx, ny))
Mag = np.ones((nx, ny))
Ang = np.zeros((nx, ny))

# 3. Structure the data to match the uploaded format:
# We stack components so that for each pixel we have [Ex, Ey, Mag, Ang]
# Shape will be (1000, 1000, 4)
field_map = np.stack([Ex, Ey, Mag, Ang], axis=-1)

# 4. Flatten the data into a single long sequence
flattened_data = field_map.flatten()

# 5. Prepend the dimensions (nx, ny) to the start of the data
final_output = np.concatenate([np.array([nx, ny]), flattened_data])

# 6. Save as a comma-separated CSV file (single row)
# 'fmt' determines the decimal precision; 'newline' ensures it stays in one row
np.savetxt(f'Emask_{nx}_uniform.csv', final_output, delimiter=',', fmt='%.1f', newline=',')

print(f"Successfully generated 'Emask1000_uniform.csv' with size {nx}x{ny}.")
```
